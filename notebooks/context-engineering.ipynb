# 上下文工程：选择、压缩、缓存与证据保真

用有限预算组织任务、工具观察和证据，判断哪些信息能压缩、哪些状态必须保持精确。

Python 3.10+ / CPU / 无密钥。固定教学 fixture，不调用模型或服务。先读对应正文，预测成功与失败，再运行并修改一个边界。

```python
def pack(items,budget):
    if any(type(x['cost']) is not int or x['cost']<0 for x in items): raise ValueError('invalid cost')
    mandatory=[x for x in items if x['required']]
    used=sum(x['cost'] for x in mandatory)
    if used>budget: raise ValueError('mandatory context exceeds budget')
    kept=mandatory[:]
    for x in sorted((x for x in items if not x['required']),
                    key=lambda x:(-x['priority'],x['id'])):
        if used+x['cost']<=budget:
            kept.append(x);used+=x['cost']
    return kept,used
items=[{'id':'goal','cost':3,'required':True,'priority':9},
       {'id':'unknown-write','cost':2,'required':True,'priority':9},
       {'id':'evidence','cost':4,'required':False,'priority':8},
       {'id':'old-log','cost':6,'required':False,'priority':1}]
selected,used=pack(items,9)
assert [x['id'] for x in selected]==['goal','unknown-write','evidence']
try: pack(items,4)
except ValueError: pass
else: raise AssertionError('mandatory information silently dropped')
checks={'mandatory_preserved':True,'budget_bound':used<=9,'overflow_explicit':True}
print([x['id'] for x in selected],used)
```
